```python
name = input("Enter your name: ")
age = int(input("Enter your age: "))

if (age < 0):
    print ("Who do you think you are? Benjamin Buttons?")

if (age >= 0) and (age <= 12):
    print ("You're young, " + name + ", get off this computer and go touch grass.")

if (age > 12) and (age <= 50):
    print ("You're healthy, " + name + ", watch for impending mid life crisis.")

if (age > 50):
    print ("You're old as hell, " + name + ", death is innevitable.")
```
